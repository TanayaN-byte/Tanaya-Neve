```python
df["Nation"] = df["Nation"].str.replace(r"\[.*?\]", "", regex=True).str.strip()
```

```python
df = df.drop_duplicates(subset="Nation", keep="first")
```

```python
top10 = df.sort_values("Total", ascending=False).head(10)
top10
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.bar(top10["Nation"], top10["Total"])

plt.xlabel("Country")
plt.ylabel("Total Medals")
plt.title("Top 10 Countries by Total Olympic Medals")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```
